```python
winning_boards = [((0,0), (0,1), (0,2)),
                    ((1,0), (1,1), (1,2)),
                    ((2,0), (2,1), (2,2)),

                    ((0,0), (1,0), (2,0)),
                    ((0,1), (1,1), (2,1)),
                    ((0,2), (1,2), (2,2)),

                    ((0,0), (1,1), (2,2)),
                    ((0,2), (1,1), (2,0)),                    
                ]

board_x_won = [ ["x", "x", "o"], 
                ["x", None, "o"], 
                ["x", "o", None]
                ]

board_draw = [  ["x", "x", "o"], 
                ["o", "o", "x"], 
                ["x", "o", "x"]
            ]

board_continue = [  ["x", "x", "o"], 
                    ["o", "o", "x"], 
                    ["x", None, "x"]
                ]

test_boards = [board_x_won, board_draw, board_continue ]

def check_if_won(board_state: list[list[str|None]], symbol_to_check: str) -> bool:

    for winning_board in winning_boards:
        has_won = True

        for x, y in winning_board: #((0,2), (1,1), (2,0))
            if board_state[x][y] != symbol_to_check:
                has_won = False
                break

        if has_won:
            return True
    
    return False

for test_board in test_boards:
    print(check_if_won(test_board, "x"))
                





```
