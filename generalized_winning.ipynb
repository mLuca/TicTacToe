```python
def check_if_won_extended(board_state: list[list], last_move_position: tuple[int, int], symbol_to_check: str, winning_streak:int = 3)-> bool:
    
    def count_streak(position: tuple[int,int], direction: tuple[int,int]) -> int:
        '''Recursively go along direction and increment count until a wrong symbol is encountered 
        or we reached limit of the board'''

        # Move the cursor
        new_position = tuple((position[0] + direction[0], position[1] + direction[1]))

        #Check if cursor still in board:
        # in x direction
        if new_position[0] > len(board_state[0]) - 1 or new_position[0] < 0:
            return 0
        # in y direction
        if new_position[1] > len(board_state) - 1  or new_position[1] < 0:
            return 0

        # check if next positon is still part of a streak
        if board_state[new_position[0]][new_position[1]] == symbol_to_check:
            return count_streak(new_position, direction) + 1
        return 0

    
    # define direction vectors to check per orientation
    # e.g. "horizontal" goes left and right, "vertical" goes up and down, ...
    orientations = {"vertical": [(0,1), (0,-1)],
                    "horizontal": [(1,0), (-1,0)],
                    "slash": [(1,1), (-1,-1)],
                    "backslash": [(1,-1), (-1,1)]}

    # do the checking
    for orientation in orientations.values():
        streak_count = 1
        for direction in orientation:
            streak_count += count_streak(last_move_position, direction)
        if streak_count >= winning_streak:
            return True

    return False

# SIMPLE TESTS
board_win_vertical = [  ["x", "x", "o"], 
                        ["x", None, "o"], 
                        ["x", "o", None]
                ]

board_win_slash = [ ["x", "o", "x"], 
                    ["x", "x", "o"], 
                    [ "o", "o", "x"]
                ]

board_no_win_1 = [  ["x", "x", "o"], 
                    ["o", None, "x"], 
                    ["x", "o", "x"]
            ]

board_no_win_2 = [  ["x", "x", None], 
                    [None, "o", "x"], 
                    ["x", None, "x"]
                ]

test_boards = [board_win_vertical, board_win_slash, board_no_win_1, board_no_win_2]


for test_board in test_boards:
    print(check_if_won_extended(test_board, (0,0), "x"))
```
